### Train-Test Split

```python
cat_cols = ['Sex', 'Embarked', 'Desk']
num_cols = ['Age', 'Fare', 'SibSp', 'Parch', 'Pclass', 'HasCabin']
```

```python
from sklearn.model_selection import train_test_split

X=df.drop('Survived',axis=1);
y=df['Survived'];

X_train,X_test,y_train,y_test=train_test_split(
    X,y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_test.shape,y_train.shape,X_test.shape,y_test.shape
```

#### divide into Categories and numerical

```python
train_cat = X_train[cat_cols]
test_cat = X_test[cat_cols]
```

```python
train_num=X_train[num_cols];
test_num=X_test[num_cols];
```

##### Encoding with One hot Encoding 

```python
from sklearn.preprocessing import OneHotEncoder;

one_hot=OneHotEncoder(sparse_output=False,drop='first');

one_hot = OneHotEncoder(
    sparse_output=False,
    drop='first',
    handle_unknown='ignore'
)

train_cat_hot = one_hot.fit_transform(train_cat)
test_cat_hot = one_hot.transform(test_cat)
```

```python
train_cat_hot
```

### Handel numerical values with scalling

```python
from sklearn.preprocessing import StandardScaler;

scaler=StandardScaler();

train_num_scale=scaler.fit_transform(train_num)
test_num_scale=scaler.transform(test_num)
```

```python
train_num_scale
```

- concatenate all categorical and numerical train
- concatenate all categorical and numerical test

```python
X_train_final = np.concatenate([train_cat_hot, train_num_scale], axis=1)
X_test_final = np.concatenate([test_cat_hot, test_num_scale], axis=1)
```

- KNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn= KNeighborsClassifier(n_neighbors=5);
knn.fit(X_train_final,y_train)
```

#### Calculate Accuarcy on Every Test

```python
from sklearn.metrics import accuracy_score

values=[3,5,7,9];

for k in values:
    knn=KNeighborsClassifier(k);
    knn.fit(X_train_final,y_train);

    y_pred=knn.predict(X_test_final);
    print(f"k = {k}, Accuracy = {accuracy_score(y_test, y_pred)*100:.4f} %")
```

### Cross Validation

```python
from sklearn.model_selection import cross_val_score

k_values = [3, 5, 7, 9]

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)

    scores = cross_val_score(
        knn, X_train_final, y_train,
        cv=5, scoring='accuracy'
    )

    print(f"k = {k}, Mean Accuracy = {scores.mean():.4f}")
```

```python
best_k=5

best_knn=KNeighborsClassifier(n_neighbors=best_k);
best_knn.fit(X_train_final,y_train);

y_pred=best_knn.predict(X_test_final);
```

### Accuracy

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("Accuracy:", accuracy_score(y_test, y_pred)*100)
print("Precision:", precision_score(y_test, y_pred)*100)
print("Recall:", recall_score(y_test, y_pred)*100)
print("F1-score:", f1_score(y_test, y_pred)*100)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

- Decesion Tree

```python
from sklearn.tree import DecisionTreeClassifier

d_tree=DecisionTreeClassifier(random_state=42);
d_tree.fit(X_train_final,y_train);

y_pred_tree=d_tree.predict(X_test_final);
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_tree)*100)
print("Precision:", precision_score(y_test, y_pred_tree)*100)
print("Recall:", recall_score(y_test, y_pred_tree)*100)
print("F1-score:", f1_score(y_test, y_pred_tree)*100)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_tree))
```

- Random Forest

```python
from sklearn.ensemble import RandomForestClassifier;

rf=RandomForestClassifier(random_state=42);
rf.fit(X_train_final,y_train);
y_pred_rf=rf.predict(X_test_final)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_rf)*100)
print("Precision:", precision_score(y_test, y_pred_rf)*100)
print("Recall:", recall_score(y_test, y_pred_rf)*100)
print("F1-score:", f1_score(y_test, y_pred_rf)*100)

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
```

#### Comaparision between 3 Algorithms
    1 KNN
    2 Decision Tree
    3 Random Forest

```python
import matplotlib.pyplot as plt
import numpy as np

metrics = ['Accuracy', 'Precision', 'Recall', 'F1-score']

knn = [76.54, 71.43, 65.22, 68.18]
decision_tree = [79.89, 75.38, 71.01, 73.13]
random_forest = [78.77, 73.85, 69.57, 71.64]

x = np.arange(len(metrics))
width = 0.25

plt.figure(figsize=(11, 6))

bars1 = plt.bar(x - width, knn, width, label='KNN')
bars2 = plt.bar(x, decision_tree, width, label='Decision Tree')
bars3 = plt.bar(x + width, random_forest, width, label='Random Forest')

plt.xlabel('Evaluation Metrics')
plt.ylabel('Score (%)')
plt.title('Model Performance Comparison')

plt.xticks(x, metrics)
plt.ylim(0, 100)

plt.legend()
plt.grid(axis='y', alpha=0.3)

for bars in [bars1, bars2, bars3]:
    plt.bar_label(bars, fmt='%.2f', padding=3)

plt.tight_layout()
plt.show()
```

### Part F - Model Comparison

Three classification models were evaluated on the Titanic test dataset: KNN, Decision Tree, and Random Forest.

The Decision Tree model achieved the best performance, with an accuracy of 79.89%, precision of 75.38%, recall of 71.01%, and F1-score of 73.13%.

Although Random Forest also achieved good results, Decision Tree performed better across all four evaluation metrics.

Therefore, **Decision Tree was selected as the best-performing model** because it achieved the highest overall test performance among the three models.
